```python
import numpy as np
import pandas as pd
from scipy import sparse
import scanpy as sc
import matplotlib.pyplot as plt
import seaborn as sns

from matplotlib import pyplot as plt
from py_monocle import (
    learn_graph,
    order_cells,
    compute_cell_states,
    regression_analysis,
    differential_expression_genes,
)
```

CD8 T lymphocytes only ✨

```python
cd8_states = ["CD8 T naive", "CD8 Tem", "CD8 Tem_ENTPD1high"]
adata_cd8 = adata[adata.obs["cell_state"].isin(cd8_states)].copy()

sc.pp.normalize_total(adata_cd8, target_sum=1e4)
sc.pp.log1p(adata_cd8)

# Find highly variable genes
sc.pp.highly_variable_genes(adata_cd8, n_top_genes=3000)

# **CRITICAL: Actually USE the HVGs for downstream analysis**
# Option 1: Subset to HVGs (recommended for trajectory analysis)
# Subset to HVGs
adata_cd8_hvg = adata_cd8[:, adata_cd8.var.highly_variable].copy()
# Scale HVGs
sc.pp.scale(adata_cd8_hvg, max_value=10)
# PCA on HVGs
sc.tl.pca(adata_cd8_hvg, n_comps=40, svd_solver="arpack")

sc.pp.neighbors(adata_cd8_hvg, n_neighbors=40, n_pcs=20)
sc.tl.leiden(
    adata_cd8_hvg,
    resolution=0.7
)

sc.tl.umap(adata_cd8_hvg, min_dist=0.05, spread=1)

# Visualize
sc.pl.umap(adata_cd8_hvg, color=["cell_state", "recist_v3", "leiden"], title="UMAP CD8 T cells only", size=40)

# **IMPORTANT: Use adata_cd8_hvg for all downstream trajectory analysis**
```

 Monocle3 trajectory

```python
# UMAP embedding
umap = adata_cd8_hvg.obsm["X_umap"]
# Cluster labels (convert to numeric)
clusters = adata_cd8_hvg.obs["leiden"].astype("category").cat.codes.values

# Learn principal graph
projected_points, mst, centroids = learn_graph(
    matrix=umap,
    clusters=clusters
)

# =========================
# ROOT CELL SELECTION
# =========================
naive_mask = adata_cd8_hvg.obs["cell_state"] == "CD8 T naive"

naive_umap = umap[naive_mask]
naive_centroid = naive_umap.mean(axis=0)

naive_indices = np.where(naive_mask)[0]
distances = np.linalg.norm(naive_umap - naive_centroid, axis=1)

root_cell_idx = naive_indices[np.argmin(distances)]
root_cells = np.array([root_cell_idx])

print(f"Selected root cell: {root_cell_idx}")
print(f"Root cell coordinates: {umap[root_cell_idx]}")

# =========================
# COMPUTE PSEUDOTIME
# =========================
pseudotime = order_cells(
    umap,
    centroids,
    mst=mst,
    projected_points=projected_points,
    root_cells=root_cells
)

# Store raw pseudotime
adata_cd8_hvg.obs["monocle_pseudotime"] = pseudotime

# =========================
# SCALE TO 0-1
# =========================
pseudotime_scaled = (
    pseudotime - pseudotime.min()
) / (
    pseudotime.max() - pseudotime.min()
)

adata_cd8_hvg.obs["monocle_pseudotime_scaled"] = pseudotime_scaled

# =========================
# PLOT TRAJECTORY
# =========================
fig, ax = plt.subplots(figsize=(5, 4))

# Extract MST edges
edges = np.array(mst.nonzero()).T

# Scatter plot
scatter = ax.scatter(
    umap[:, 0],
    umap[:, 1],
    c=pseudotime_scaled,
    s=4,
    cmap="GnBu",
    vmin=0,
    vmax=1,
    alpha=1,
    rasterized=True
)

# Plot MST edges
for edge in edges:
    ax.plot(
        centroids[edge, 0],
        centroids[edge, 1],
        c="red",
        linewidth=0.5,
        alpha=0.9
    )

# Styling
ax.set_title("Pseudotime trajectory (scaled)", fontsize=14)

ax.set_xticks([])
ax.set_yticks([])

for spine in ax.spines.values():
    spine.set_visible(False)

# Colorbar
cbar = plt.colorbar(
    scatter,
    ax=ax,
    label="Scaled pseudotime (0–1)"
)
cbar.ax.tick_params(labelsize=10)

plt.tight_layout()

plt.savefig(
    "./figures/monocle/pseudotime_trajectory_scaled.pdf",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# QUICK CHECK
# =========================
print(
    f"Raw pseudotime range: "
    f"{pseudotime.min():.3f} - {pseudotime.max():.3f}"
)

print(
    f"Scaled pseudotime range: "
    f"{pseudotime_scaled.min():.3f} - {pseudotime_scaled.max():.3f}"
)
```

```python
# =========================
# Visualize results
# =========================
import matplotlib.pyplot as plt
import scanpy as sc

fig, axes = plt.subplots(1, 4, figsize=(20, 4))

# -------------------------
# Panel A: Cell states
# -------------------------
sc.pl.umap(
    adata_cd8_hvg,
    color="cell_state",
    ax=axes[0],
    show=False,
    title="Cell states",
    legend_loc="on data"
)

# -------------------------
# Panel B: RECIST
# -------------------------
sc.pl.umap(
    adata_cd8_hvg,
    color="recist_v3",
    ax=axes[1],
    show=False,
    title="RECIST"
)

# -------------------------
# Panel C: Scaled pseudotime
# -------------------------
sc.pl.umap(
    adata_cd8_hvg,
    color="monocle_pseudotime_scaled",
    ax=axes[2],
    show=False,
    cmap="GnBu",
    vmin=0,
    vmax=1,
    title="Scaled pseudotime (0–1)"
)

# -------------------------
# Panel D: Root cell
# -------------------------
axes[3].scatter(
    umap[:, 0],
    umap[:, 1],
    c="lightgray",
    s=8,
    alpha=0.5,
    rasterized=True
)

axes[3].scatter(
    umap[root_cell_idx, 0],
    umap[root_cell_idx, 1],
    c="red",
    s=250,
    marker="*",
    label="Root cell",
    edgecolors="black",
    linewidths=1.5
)

axes[3].set_title("Root cell location")
axes[3].legend(frameon=False)

axes[3].set_xticks([])
axes[3].set_yticks([])

for spine in axes[3].spines.values():
    spine.set_visible(False)

plt.tight_layout()

plt.savefig(
    "./figures/monocle/pseudotime_summary_panels.pdf",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# Verify pseudotime progression
# =========================

print("\nMedian scaled pseudotime by cell state:")
print(
    adata_cd8_hvg.obs.groupby("cell_state")[
        "monocle_pseudotime_scaled"
    ].median().sort_values()
)

print("\nMean scaled pseudotime by cell state:")
print(
    adata_cd8_hvg.obs.groupby("cell_state")[
        "monocle_pseudotime_scaled"
    ].mean().sort_values()
)

print("\nFull distribution by cell state:")
print(
    adata_cd8_hvg.obs.groupby("cell_state")[
        "monocle_pseudotime_scaled"
    ].describe()
)
```

```python
# Order states biologically
state_order = ["CD8 T naive","CD8 Tem_ENTPD1high", "CD8 Tem" ]

# Create figure
fig, ax = plt.subplots(figsize=(12, 4))

# Vertical spacing between ridges
y_offsets = np.arange(len(state_order))

for i, state in enumerate(state_order):

    mask = adata_cd8_hvg.obs["cell_state"] == state
    pseudotime_vals = adata_cd8_hvg.obs.loc[mask, "monocle_pseudotime"].values

    # Compute KDE (invisible, used only to extract curve)
    kde = sns.kdeplot(
        x=pseudotime_vals,
        bw_adjust=0.8,
        fill=False,
        ax=ax,
        color="none",   # ← prevents unwanted visible line
        linewidth=0
    )

    # Extract curve
    line = kde.get_lines()[-1]
    x, y = line.get_data()

    # Normalize height
    y = y / y.max() * 0.9

    # Draw filled ridge ONLY
    ax.fill_between(
        x,
        y_offsets[i],
        y + y_offsets[i],
        color=cell_state_colors[state],
        alpha=0.9,
        linewidth=0   # ← removes edge line
    )

# Formatting
ax.set_yticks(y_offsets)
ax.set_yticklabels(state_order, fontsize=11)
ax.set_xlabel("Monocle pseudotime", fontsize=12)
ax.set_ylabel("")
ax.set_title("CD8 T cell trajectory progression", fontsize=13)

# Clean Nature-style axes
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig("./figures/monocle/cd8_pseudotime_kde_vertical.pdf", dpi=300, bbox_inches='tight')
plt.show()
```
